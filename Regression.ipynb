```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Feature_engineering_done.csv")
df
```

```python
df = df.drop(columns=['Unnamed: 0.1','Unnamed: 0'], axis=1)
```

```python
print(df.shape)
```

```python
x = df.drop(columns=['salary_usd'])
y = df['salary_usd']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
Y_train
```

```python
X_train.shape
```

```python
Y_train.shape
```

```python
from sklearn.linear_model import LinearRegression

linear = LinearRegression()
linear.fit(X_train,Y_train)
Y_pred = linear.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score , root_mean_squared_error

mse1 = mean_squared_error(Y_test,Y_pred)
rmse1 = root_mean_squared_error(Y_test,Y_pred)
mae1 = mean_absolute_error(Y_test,Y_pred)
r2_score1 = r2_score(Y_test,Y_pred)
```

```python
print("mean_squared_error : ",mse1)
print("root_mean_squared_error : ",rmse1)
print("mean_absolute_error : ",mae1)
print("r2_score : ",r2_score1)
```

## Decission Tree

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dt = DecisionTreeRegressor(random_state=42,max_depth=10)
dt.fit(X_train, Y_train)
y_pred = dt.predict(X_test)
```

```python
mse2 = mean_squared_error(Y_test,Y_pred)
rmse2 = root_mean_squared_error(Y_test,Y_pred)
mae2 = mean_absolute_error(Y_test,Y_pred)
r2_score2 = r2_score(Y_test,Y_pred)
```

```python
print("mean_squared_error : ",mse2)
print("root_mean_squared_error : ",rmse2)
print("mean_absolute_error : ",mae2)
print("r2_score : ",r2_score2)
```

## SVR

```python
from sklearn.svm import SVR

svr = SVR(kernel='rbf', C=100, epsilon=0.1)
svr.fit(X_train, Y_train)

y_pred = svr.predict(X_test)
```

```python
mse3 = mean_squared_error(Y_test,Y_pred)
rmse3 = root_mean_squared_error(Y_test,Y_pred)
mae3 = mean_absolute_error(Y_test,Y_pred)
r2_score3 = r2_score(Y_test,Y_pred)
```

```python
print("mean_squared_error : ",mse3)
print("root_mean_squared_error : ",rmse3)
print("mean_absolute_error : ",mae3)
print("r2_score : ",r2_score3)
```
