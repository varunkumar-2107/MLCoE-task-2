```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Feature_engineering_done.csv")
df
```

```python
df = df.drop(columns=['Unnamed: 0.1','Unnamed: 0'], axis=1)
```

```python
print(df.shape)
```

```python
x = df.drop(columns=['salary_usd'])
y = df['salary_usd']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
Y_train
```

```python
X_train.shape
```

## Linear Regression


```python
from sklearn.linear_model import LinearRegression

linear = LinearRegression()
linear.fit(X_train,Y_train)
Y_pred = linear.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score , root_mean_squared_error

mse1 = mean_squared_error(Y_test,Y_pred)
rmse1 = root_mean_squared_error(Y_test,Y_pred)
mae1 = mean_absolute_error(Y_test,Y_pred)
r2_score1 = r2_score(Y_test,Y_pred)
```

```python
print("mean_squared_error : ",mse1)
print("root_mean_squared_error : ",rmse1)
print("mean_absolute_error : ",mae1)
print("r2_score : ",r2_score1)
```

## Short Note on Linear regression

To train a Linear regression model it have two method:

1. OLS (Ordinary Least Squares) method : It is a method in which coeff of the equation is obtain directly by the formula 
- For linear feature let Eq is :  Y  = mx + b
so the Formula is =>    ( b = Y - m(xm) )  &  ( m = sum of rows((x-xm)(y-ym)) / sum(x - xm)^2 )
where : Y = predicted value , xm = mean of imput feature , ym = mean of target feature
- For Multiple Feature :  formula =>  B = (X.T . X)^-1 (X.T).Y
Where : B = matrix of all the coeff of feature and intecept,  X = Matrix of values of all the feature , Y = matrix of Value of a target column 



2. Gradient descent method : It is the method in which we have to minimize the loss function by using gradient descent function we try multple iteraction to minimize the loss funtion
- Loss Function = (mean squared error) = 1/n( sum all row value( (y-Y)^2 ) )
where Y = predicted value , y = target value , n = no of rows
- Gradient descent = Qnew = Qold - a(dL/dQold)
where Q = Coeff of any feature  , L = Loss function
