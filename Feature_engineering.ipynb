```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("EDA_done_data.csv")
df.head(10).T
```

```python
df.select_dtypes(include=['int','float']).columns
```

```python
df['salary_usd'] = (df['salary_min_usd']+df['salary_max_usd'])/2
```

```python
obj_col = df.select_dtypes(include=["object"])
for col in obj_col:
    print(f"{col}  =>  {df[col].nunique()}")
```

```python
df["posting_date"] = pd.to_datetime(df["posting_date"],errors='coerce')
df["application_deadline"] = pd.to_datetime(df["application_deadline"],errors='coerce')
df["date_position_filled"] = pd.to_datetime(df["date_position_filled"],errors='coerce')
```

```python
df[df["posting_date"] > df["application_deadline"]]
```

```python
df[df["date_position_filled"] > df["posting_date"]]
```

```python
df["Application_active_days"] = (df['application_deadline'] - df["posting_date"]).dt.days
```

```python
df.head(10).T
```

```python
df.isnull().sum()
```

```python
df['Application_active_days'] = df['Application_active_days'].fillna(-1)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['job_id','company_name','required_tech_stack','job_location_city','salary_min_usd','salary_max_usd','application_deadline','date_position_filled','posting_date'])
```

```python
obj_col = df.select_dtypes(include=["object"])
for col in obj_col:
    print(f"{col}  =>  {df[col].unique()}")
    print(" ")
```

```python
def abc(str):
    str = str.strip()
    str = str.lower()
    return str

df['job_role'] = df['job_role'].apply(abc)
df['sector'] = df['sector'].apply(abc)
df['work_mode'] = df['work_mode'].apply(abc)

# df['job_role'].unique()
```

```python
df['work_mode'] = df['work_mode'].replace({'wfh':'work from home', 'on-site':'onsite'})
df['work_mode'].unique()
```

```python
obj_col = df.select_dtypes(include=["object"])
for col in obj_col:
    print(f"{col}  =>  {df[col].nunique()}")
```

```python
# One hot encoding
df = pd.get_dummies(df,columns=['company_type','sector','hq_country','seniority_level','work_mode','job_location_country','primary_programming_language','salary_currency','position_status','job_role'], dtype=int, drop_first=True)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['No','Yes']], dtype=int)
df['visa_sponsorship_offered'] = oe.fit_transform(df[['visa_sponsorship_offered']])
df['layoff_within_12_months_flag'] = oe.fit_transform(df[['layoff_within_12_months_flag']])
```
